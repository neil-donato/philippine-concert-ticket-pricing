```python
from pathlib import Path

import joblib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import RandomForestRegressor
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import GroupShuffleSplit
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

DATA_PATH = Path('../data/philippine_concert_tickets.csv')
MODEL_PATH = Path('../models/concert_ticket_price_model.pkl')
TARGET = 'ticket_price_php'
RANDOM_STATE = 42
sns.set_theme(style='whitegrid')


def show(title, table):
    print(f'\n{title}')
    print(table.to_string(index=False) if isinstance(table, pd.DataFrame) else table)


def grouped_summary(df, group_col, stats=('count', 'mean', 'median')):
    return (df.groupby(group_col, dropna=False)[TARGET]
            .agg(list(stats)).sort_values('mean', ascending=False).reset_index())


def bar_chart(table, category, title, top=None):
    plot_df = table.head(top) if top else table
    if plot_df.empty:
        return
    plt.figure(figsize=(12, max(5, min(12, len(plot_df) * .4))))
    sns.barplot(data=plot_df, x='mean', y=category, color='steelblue')
    plt.title(title)
    plt.xlabel('Average ticket price (PHP)')
    plt.ylabel(category.replace('_', ' ').title())
    plt.tight_layout()
    plt.show()


def load_and_clean():
    df = pd.read_csv(DATA_PATH)
    show('RAW DATA PREVIEW', df.head())
    print(f'Original shape: {df.shape}; missing cells: {df.isna().sum().sum()}; duplicate rows: {df.duplicated().sum()}')
    show('NUMERICAL SUMMARY', df.describe().T)

    clean = df.copy()
    clean[TARGET] = pd.to_numeric(clean[TARGET], errors='coerce')
    clean = clean.dropna(subset=[TARGET]).copy()
    if 'concert_date' in clean:
        clean['concert_date'] = pd.to_datetime(clean['concert_date'], errors='coerce')
    for col in clean.select_dtypes(include=['object', 'string']).columns:
        clean[col] = clean[col].astype('string').str.strip()
    clean = clean.drop_duplicates().reset_index(drop=True)
    if 'concert_date' in clean:
        clean['concert_year'] = clean['concert_date'].dt.year
        clean['concert_month'] = clean['concert_date'].dt.month
        clean['concert_day_of_week'] = clean['concert_date'].dt.day_name()
    print(f'Cleaned shape: {clean.shape}; rows removed: {len(df) - len(clean)}')
    show('REMAINING MISSING VALUES', clean.isna().sum())
    return clean


def explore(df):
    prices = df[TARGET]
    show('TICKET PRICE SUMMARY', prices.describe())
    print(f'Median: PHP {prices.median():,.2f} | Mean: PHP {prices.mean():,.2f}')

    plt.figure(figsize=(10, 5))
    sns.histplot(prices, bins=40, kde=True, color='steelblue')
    plt.title('Distribution of Concert Ticket Prices in the Philippines')
    plt.xlabel('Ticket price (PHP)')
    plt.ylabel('Ticket-tier records')
    plt.tight_layout()
    plt.show()

    plt.figure(figsize=(10, 3))
    sns.boxplot(x=prices, color='skyblue')
    plt.title('Boxplot of Concert Ticket Prices')
    plt.xlabel('Ticket price (PHP)')
    plt.tight_layout()
    plt.show()

    for column, title in [('artist_category', 'Average Ticket Price by Artist Category'),
                          ('ticket_tier', 'Average Ticket Price by Ticket Tier')]:
        if column not in df:
            continue
        stats = ('count', 'mean', 'median', 'min', 'max') if column == 'ticket_tier' else ('count', 'mean', 'median')
        summary = grouped_summary(df, column, stats)
        show(title.upper(), summary.round(2))
        bar_chart(summary, column, title)
        if column == 'ticket_tier':
            plt.figure(figsize=(12, 6))
            sns.boxplot(data=df, x=TARGET, y=column, order=summary[column],
                        color='skyblue', showfliers=False)
            plt.title('Ticket Price Distribution by Ticket Tier')
            plt.xlabel('Ticket price (PHP)')
            plt.tight_layout()
            plt.show()

    if 'concert_year' in df:
        yearly = (df.dropna(subset=['concert_year']).groupby('concert_year')[TARGET]
                  .agg(['count', 'mean', 'median']).reset_index().sort_values('concert_year'))
        show('YEARLY TICKET PRICE SUMMARY', yearly.round(2))
        if not yearly.empty:
            plt.figure(figsize=(10, 5))
            sns.lineplot(data=yearly, x='concert_year', y='mean', marker='o', color='steelblue')
            for _, row in yearly.iterrows():
                plt.annotate(f"₱{row['mean']:,.0f}", (row['concert_year'], row['mean']),
                             xytext=(0, 10), textcoords='offset points', ha='center')
            plt.title('Average Concert Ticket Prices by Year')
            plt.xticks(yearly['concert_year'])
            plt.xlabel('Concert year')
            plt.ylabel('Average ticket price (PHP)')
            plt.tight_layout()
            plt.show()

    if 'venue' in df:
        venue_summary = grouped_summary(df, 'venue')
        show('TOP 10 VENUES BY AVERAGE PRICE', venue_summary.head(10).round(2))
        bar_chart(venue_summary, 'venue', 'Top 10 Venues by Average Ticket Price', top=10)

    location = next((c for c in ['city_or_province', 'city', 'location'] if c in df), None)
    if location:
        location_summary = grouped_summary(df, location)
        show('TICKET PRICES BY LOCATION', location_summary.round(2))
        bar_chart(location_summary, location, 'Average Ticket Price by Location')

    popularity = next((c for c in ['artist_popularity', 'artist_popularity_score',
                                    'popularity_score', 'popularity_index'] if c in df), None)
    if popularity:
        pop_df = df[[popularity, TARGET]].copy()
        pop_df[popularity] = pd.to_numeric(pop_df[popularity], errors='coerce')
        pop_df = pop_df.dropna()
        if len(pop_df) >= 2 and pop_df[popularity].nunique() > 1:
            print(f'Artist popularity / price Pearson correlation: {pop_df[popularity].corr(pop_df[TARGET]):.4f}')
            plt.figure(figsize=(11, 6))
            sns.regplot(data=pop_df, x=popularity, y=TARGET,
                        scatter_kws={'alpha': .2, 's': 20}, line_kws={'color': 'red'})
            plt.title('Artist Popularity vs Ticket Price')
            plt.xlabel('Artist popularity score')
            plt.ylabel('Ticket price (PHP)')
            plt.tight_layout()
            plt.show()

    numeric = df.select_dtypes(include='number').copy()
    numeric = numeric.drop(columns=[c for c in numeric if c.lower() == 'id' or c.lower().endswith('_id')])
    numeric = numeric.loc[:, numeric.nunique() > 1]
    if numeric.shape[1] >= 2:
        corr = numeric.corr()
        show('CORRELATION MATRIX', corr.round(3))
        plt.figure(figsize=(12, 8))
        sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', center=0,
                    vmin=-1, vmax=1, linewidths=.5)
        plt.title('Correlation Heatmap of Concert Pricing Variables')
        plt.tight_layout()
        plt.show()
        if TARGET in corr:
            show('CORRELATIONS WITH TICKET PRICE', corr[TARGET].drop(TARGET).sort_values(ascending=False).round(3))


def model_and_evaluate(df):
    candidate_features = ['artist_category', 'ticket_tier', 'venue', 'city_or_province',
                          'artist_popularity', 'artist_popularity_score', 'popularity_score',
                          'popularity_index', 'venue_capacity', 'concert_year', 'concert_month']
    features = [col for col in candidate_features if col in df]
    if not features:
        raise ValueError('No expected predictor columns found.')
    if 'concert_id' in df:
        groups = df['concert_id'].astype('string').fillna('missing')
    elif {'concert_name', 'concert_date', 'venue'}.issubset(df.columns):
        groups = (df['concert_name'].astype('string').fillna('missing') + '|' +
                  df['concert_date'].astype('string').fillna('missing') + '|' +
                  df['venue'].astype('string').fillna('missing'))
    else:
        raise ValueError('Concert identifier or concert name/date/venue required for group split.')
    if groups.nunique() < 2:
        raise ValueError('At least two distinct concert groups are required.')

    X, y = df[features], df[TARGET]
    split = GroupShuffleSplit(n_splits=1, test_size=.2, random_state=RANDOM_STATE)
    train_idx, test_idx = next(split.split(X, y, groups=groups))
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]
    print(f'Features: {features}')
    print(f'Train/test records: {len(X_train)}/{len(X_test)}')
    print(f'Train/test concerts: {groups.iloc[train_idx].nunique()}/{groups.iloc[test_idx].nunique()}')

    categorical = X_train.select_dtypes(include=['object', 'string', 'category']).columns.tolist()
    numerical = X_train.select_dtypes(include='number').columns.tolist()

    def make_preprocessor():
        return ColumnTransformer([
            ('num', Pipeline([('imputer', SimpleImputer(strategy='median')),
                              ('scaler', StandardScaler())]), numerical),
            ('cat', Pipeline([('imputer', SimpleImputer(strategy='most_frequent')),
                              ('encoder', OneHotEncoder(handle_unknown='ignore'))]), categorical),
        ])

    models = {
        'Linear Regression': Pipeline([('preprocessor', make_preprocessor()),
                                       ('regressor', LinearRegression())]),
        'Random Forest': Pipeline([('preprocessor', make_preprocessor()),
                                   ('regressor', RandomForestRegressor(
                                       n_estimators=100, max_depth=15, min_samples_leaf=2,
                                       random_state=RANDOM_STATE, n_jobs=-1))]),
    }
    predictions, metrics = {}, []
    for name, model in models.items():
        model.fit(X_train, y_train)
        pred = model.predict(X_test)
        predictions[name] = pred
        metrics.append({'Model': name, 'MAE': mean_absolute_error(y_test, pred),
                        'RMSE': np.sqrt(mean_squared_error(y_test, pred)),
                        'R2': r2_score(y_test, pred)})
        show(f'{name.upper()} - ACTUAL VS PREDICTED', pd.DataFrame({
            'Actual Price': y_test.to_numpy(), 'Predicted Price': pred}).head(10).round(2))
        plt.figure(figsize=(9, 6))
        plt.scatter(y_test, pred, alpha=.3, color='steelblue')
        low, high = min(y_test.min(), pred.min()), max(y_test.max(), pred.max())
        plt.plot([low, high], [low, high], 'r--', label='Perfect prediction')
        plt.title(f'{name}: Actual vs Predicted Prices')
        plt.xlabel('Actual price (PHP)')
        plt.ylabel('Predicted price (PHP)')
        plt.legend()
        plt.tight_layout()
        plt.show()

    comparison = pd.DataFrame(metrics).sort_values('MAE').reset_index(drop=True)
    show('MODEL COMPARISON', comparison.round(3))
    for metric, label in [('MAE', 'Mean Absolute Error (PHP)'),
                          ('RMSE', 'Root Mean Squared Error (PHP)')]:
        plt.figure(figsize=(8, 5))
        sns.barplot(data=comparison, x='Model', y=metric, color='steelblue')
        plt.title(f'Model Comparison: {metric}')
        plt.ylabel(label)
        plt.tight_layout()
        plt.show()

    forest = models['Random Forest']
    importance = pd.DataFrame({
        'Feature': forest.named_steps['preprocessor'].get_feature_names_out(),
        'Importance': forest.named_steps['regressor'].feature_importances_,
    }).sort_values('Importance', ascending=False)
    show('TOP 15 RANDOM FOREST FEATURES', importance.head(15).round(4))
    plt.figure(figsize=(12, 7))
    sns.barplot(data=importance.head(15), x='Importance', y='Feature', color='steelblue')
    plt.title('Top 15 Random Forest Feature Importances')
    plt.tight_layout()
    plt.show()

    best_name = comparison.iloc[0]['Model']  # lowest test MAE (exploratory selection)
    best_model = models[best_name]
    pred = predictions[best_name]
    evaluation = pd.DataFrame({'Actual Price': y_test.to_numpy(), 'Predicted Price': pred})
    evaluation['Error'] = evaluation['Actual Price'] - evaluation['Predicted Price']
    evaluation['Absolute Error'] = evaluation['Error'].abs()
    evaluation['Percentage Error'] = np.where(evaluation['Actual Price'] > 0,
                                               evaluation['Absolute Error'] / evaluation['Actual Price'] * 100,
                                               np.nan)
    print(f'Best model by test MAE: {best_name}')
    show('ERROR SUMMARY', evaluation[['Error', 'Absolute Error', 'Percentage Error']].describe().round(2))
    show('10 LARGEST ERRORS', evaluation.nlargest(10, 'Absolute Error').round(2))
    plt.figure(figsize=(10, 5))
    sns.histplot(evaluation['Error'], bins=40, kde=True, color='steelblue')
    plt.axvline(0, color='red', linestyle='--')
    plt.title('Prediction Error Distribution')
    plt.xlabel('Actual - predicted price (PHP)')
    plt.tight_layout()
    plt.show()
    plt.figure(figsize=(10, 5))
    sns.scatterplot(data=evaluation, x='Predicted Price', y='Error', alpha=.3, color='steelblue')
    plt.axhline(0, color='red', linestyle='--')
    plt.title('Residuals vs Predicted Prices')
    plt.tight_layout()
    plt.show()

    MODEL_PATH.parent.mkdir(parents=True, exist_ok=True)
    joblib.dump(best_model, MODEL_PATH)
    print(f'Saved {best_name} to {MODEL_PATH.resolve()}')
    loaded = joblib.load(MODEL_PATH)  # only load models you trust
    sample = X_test.head(5).copy()
    sample['Predicted Ticket Price (PHP)'] = loaded.predict(X_test.head(5)).round(2)
    show('SAMPLE PREDICTIONS', sample)
    print('Note: Choosing a model on the test set can bias reported performance; use validation data for a final assessment.')


def main():
    df_clean = load_and_clean()
    explore(df_clean)
    model_and_evaluate(df_clean)


if __name__ == '__main__':
    main()
```
